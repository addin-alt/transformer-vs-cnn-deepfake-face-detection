```python
# ================================================================================================
# BEYOND ATTENTION — COMPLETE CNN EXTENSION + FOUR-MODEL RESEARCH PIPELINE
# ================================================================================================
#
# PROJECT:
# Beyond Attention:
# A Cross-Architecture Study of CNNs and Vision Transformers
# for GAN-Generated Face Detection
#
# DATASET:
# xhlulu / 140k Real and Fake Faces
#
# EXISTING TRANSFORMER RESULTS REUSED:
#   ViT-Base/16
#   Swin-Tiny
#
# MODELS TRAINED IN THIS NOTEBOOK:
#   ResNet50
#   EfficientNet-B0
#
# IMPORTANT:
#   - ViT and Swin are NOT retrained.
#   - Same dataset.
#   - Same seed = 42.
#   - Same 224 x 224 policy.
#   - Same 10,000 train + 2,000 validation bake-off.
#   - Fake is the positive forensic class.
#   - Winner selected using validation fake-class ROC-AUC.
#   - If Swin remains winner, previous full Swin final result is reused.
#   - If a CNN wins, it is automatically trained on the full training set for 2 epochs.
#
# OUTPUT:
#   /kaggle/working/beyond_attention_four_model_research/
#
# ================================================================================================


# ================================================================================================
# SECTION 0 — INSTALL / IMPORT
# ================================================================================================

import sys
import subprocess

print("=" * 110)
print("SECTION 0 — DEPENDENCIES")
print("=" * 110)

subprocess.check_call([
    sys.executable,
    "-m",
    "pip",
    "install",
    "-q",
    "timm>=1.0.20,<1.1",
    "scikit-learn"
])

import os
import gc
import json
import math
import time
import random
import shutil
import hashlib
import warnings
from pathlib import Path
from collections import Counter, defaultdict

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image, ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms

import timm

from tqdm.auto import tqdm

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    cohen_kappa_score,
    matthews_corrcoef,
    confusion_matrix,
    roc_curve,
    precision_recall_curve,
    classification_report
)

from sklearn.calibration import calibration_curve

print("✅ Dependencies ready.")


# ================================================================================================
# SECTION 1 — CONFIGURATION
# ================================================================================================

SEED = 42
IMAGE_SIZE = 224

# Same controlled architecture bake-off used previously
BAKE_TRAIN_SIZE = 10_000
BAKE_VAL_SIZE = 2_000
BAKE_EPOCHS = 1

# Only used if ResNet50 or EfficientNet-B0 defeats Swin
FULL_EPOCHS = 2

TRAIN_BATCH_SIZE = 16
EVAL_BATCH_SIZE = 32
GRAD_ACCUMULATION = 1

LEARNING_RATE = 3e-5
WEIGHT_DECAY = 0.01
GRAD_CLIP_NORM = 1.0

NUM_WORKERS = 2

# Time-efficient image-quality audit.
# This DOES NOT reduce training data.
# It only limits expensive image-statistics inspection.
QUALITY_IMAGES_PER_SPLIT = 2000

OUTPUT_ROOT = Path(
    "/kaggle/working/beyond_attention_four_model_research"
)

DATA_DIR = OUTPUT_ROOT / "00_Data_Preprocessing"
RESNET_DIR = OUTPUT_ROOT / "01_ResNet50"
EFFNET_DIR = OUTPUT_ROOT / "02_EfficientNetB0"
COMPARE_DIR = OUTPUT_ROOT / "03_Four_Model_Comparison"
EXPLAIN_DIR = OUTPUT_ROOT / "04_Explainability"
FINAL_DIR = OUTPUT_ROOT / "05_Final_Winner"

if OUTPUT_ROOT.exists():
    shutil.rmtree(OUTPUT_ROOT)

for directory in [
    OUTPUT_ROOT,
    DATA_DIR,
    RESNET_DIR,
    EFFNET_DIR,
    COMPARE_DIR,
    EXPLAIN_DIR,
    FINAL_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)

print("✅ Configuration ready.")


# ================================================================================================
# SECTION 2 — REPRODUCIBILITY
# ================================================================================================

random.seed(SEED)
np.random.seed(SEED)

torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.benchmark = True

print("✅ Seed:", SEED)


# ================================================================================================
# SECTION 3 — HARDWARE
# ================================================================================================

print("\n" + "=" * 110)
print("SECTION 3 — HARDWARE")
print("=" * 110)

print("Python:", sys.version.split()[0])
print("PyTorch:", torch.__version__)
print("timm:", timm.__version__)
print("CUDA available:", torch.cuda.is_available())
print("Visible GPUs:", torch.cuda.device_count())

for gpu_id in range(torch.cuda.device_count()):
    print(f"GPU {gpu_id}: {torch.cuda.get_device_name(gpu_id)}")

if not torch.cuda.is_available():
    raise RuntimeError(
        "GPU is not enabled. "
        "Kaggle → Notebook options → Accelerator → GPU."
    )

DEVICE = torch.device("cuda:0")

print("Training device:", DEVICE)
print("✅ GPU 0 will be used.")


# ================================================================================================
# SECTION 4 — FIND DATASET
# ================================================================================================

print("\n" + "=" * 110)
print("SECTION 4 — DATASET DISCOVERY")
print("=" * 110)

USER_DATASET_ROOT = Path(
    "/kaggle/input/datasets/xhlulu/140k-real-and-fake-faces"
)


def is_dataset_root(path):

    path = Path(path)

    train_ok = (
        (path / "train" / "fake").is_dir()
        and
        (path / "train" / "real").is_dir()
    )

    test_ok = (
        (path / "test" / "fake").is_dir()
        and
        (path / "test" / "real").is_dir()
    )

    validation_ok = any(
        (path / name / "fake").is_dir()
        and
        (path / name / "real").is_dir()
        for name in ["valid", "validation", "val"]
    )

    return train_ok and validation_ok and test_ok


candidate_paths = [
    USER_DATASET_ROOT,
    USER_DATASET_ROOT / "real_vs_fake",
    USER_DATASET_ROOT / "real_vs_fake" / "real-vs-fake",
    Path(
        "/kaggle/input/140k-real-and-fake-faces/"
        "real_vs_fake/real-vs-fake"
    )
]

DATASET_ROOT = None

for candidate in candidate_paths:

    if is_dataset_root(candidate):
        DATASET_ROOT = candidate
        break


if DATASET_ROOT is None:

    print("Searching inside supplied dataset directory...")

    for root, dirs, files in os.walk(USER_DATASET_ROOT):

        root_path = Path(root)

        if is_dataset_root(root_path):
            DATASET_ROOT = root_path
            break


if DATASET_ROOT is None:

    raise FileNotFoundError(
        "Could not find train/fake, train/real, "
        "valid/fake, valid/real, test/fake and test/real."
    )


TRAIN_PATH = DATASET_ROOT / "train"
TEST_PATH = DATASET_ROOT / "test"

if (DATASET_ROOT / "valid").is_dir():
    VAL_PATH = DATASET_ROOT / "valid"

elif (DATASET_ROOT / "validation").is_dir():
    VAL_PATH = DATASET_ROOT / "validation"

else:
    VAL_PATH = DATASET_ROOT / "val"


print("✅ DATASET FOUND")
print("Root:", DATASET_ROOT)
print("Train:", TRAIN_PATH)
print("Validation:", VAL_PATH)
print("Test:", TEST_PATH)


# ================================================================================================
# SECTION 5 — RAW DATASET LOAD
# ================================================================================================

raw_train = datasets.ImageFolder(TRAIN_PATH)
raw_val = datasets.ImageFolder(VAL_PATH)
raw_test = datasets.ImageFolder(TEST_PATH)

CLASSES = raw_train.classes
CLASS_TO_ID = raw_train.class_to_idx

FAKE_ID = CLASS_TO_ID["fake"]
REAL_ID = CLASS_TO_ID["real"]

print("\n" + "=" * 110)
print("SECTION 5 — DATASET OVERVIEW")
print("=" * 110)

print("Classes:", CLASSES)
print("Class mapping:", CLASS_TO_ID)

print(f"Training images   : {len(raw_train):,}")
print(f"Validation images : {len(raw_val):,}")
print(f"Test images       : {len(raw_test):,}")
print(f"TOTAL             : {len(raw_train)+len(raw_val)+len(raw_test):,}")

print("Positive forensic class: FAKE")


# ================================================================================================
# SECTION 6 — CLASS DISTRIBUTION
# ================================================================================================

def class_counts(dataset):

    targets = np.asarray(dataset.targets)

    return {
        name: int(np.sum(targets == class_id))
        for name, class_id in dataset.class_to_idx.items()
    }


train_counts = class_counts(raw_train)
val_counts = class_counts(raw_val)
test_counts = class_counts(raw_test)

print("\nTraining:", train_counts)
print("Validation:", val_counts)
print("Test:", test_counts)

distribution_rows = []

for split, counts in [
    ("Train", train_counts),
    ("Validation", val_counts),
    ("Test", test_counts)
]:

    for class_name in CLASSES:

        distribution_rows.append({
            "split": split,
            "class": class_name,
            "count": counts[class_name]
        })


distribution_df = pd.DataFrame(distribution_rows)

distribution_df.to_csv(
    DATA_DIR / "dataset_class_distribution.csv",
    index=False
)


fig, ax = plt.subplots(figsize=(9, 5))

splits = ["Train", "Validation", "Test"]

x = np.arange(len(splits))

width = 0.35

for class_number, class_name in enumerate(CLASSES):

    values = []

    for split in splits:

        value = distribution_df[
            (distribution_df["split"] == split)
            &
            (distribution_df["class"] == class_name)
        ]["count"].iloc[0]

        values.append(value)

    bars = ax.bar(
        x + (class_number - 0.5) * width,
        values,
        width,
        label=class_name.title()
    )

    for bar, value in zip(bars, values):

        ax.text(
            bar.get_x() + bar.get_width()/2,
            value,
            f"{int(value):,}",
            ha="center",
            va="bottom",
            fontsize=9
        )


ax.set_xticks(x)
ax.set_xticklabels(splits)
ax.set_ylabel("Number of Images")
ax.set_title("Dataset Distribution Across Official Splits")
ax.legend()

fig.tight_layout()

fig.savefig(
    DATA_DIR / "01_dataset_distribution.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 7 — DATA LEAKAGE CHECK BY FILE NAMES
# ================================================================================================

print("\n" + "=" * 110)
print("SECTION 7 — BASIC SPLIT LEAKAGE CHECK")
print("=" * 110)


def basename_set(dataset):

    return {
        Path(path).name
        for path, _
        in dataset.samples
    }


train_names = basename_set(raw_train)
val_names = basename_set(raw_val)
test_names = basename_set(raw_test)

train_val_overlap = train_names.intersection(val_names)
train_test_overlap = train_names.intersection(test_names)
val_test_overlap = val_names.intersection(test_names)

leakage_df = pd.DataFrame([
    {
        "comparison": "Train vs Validation",
        "duplicate_filenames": len(train_val_overlap)
    },
    {
        "comparison": "Train vs Test",
        "duplicate_filenames": len(train_test_overlap)
    },
    {
        "comparison": "Validation vs Test",
        "duplicate_filenames": len(val_test_overlap)
    }
])

print(leakage_df.to_string(index=False))

leakage_df.to_csv(
    DATA_DIR / "split_filename_overlap.csv",
    index=False
)


# ================================================================================================
# SECTION 8 — QUALITY-AUDIT SAMPLE
# ================================================================================================

print("\n" + "=" * 110)
print("SECTION 8 — IMAGE QUALITY / PREPROCESSING AUDIT")
print("=" * 110)

print(
    "Training still uses the full official dataset.\n"
    "For expensive image-quality statistics, a reproducible sample "
    "is inspected to stay inside the assignment time limit."
)


def stratified_audit_indices(dataset, requested):

    targets = np.asarray(dataset.targets)

    indices = np.arange(len(dataset))

    if requested >= len(dataset):
        return indices

    selected, _ = train_test_split(
        indices,
        train_size=requested,
        random_state=SEED,
        stratify=targets
    )

    return np.asarray(selected)


audit_sets = {
    "Train": (
        raw_train,
        stratified_audit_indices(
            raw_train,
            QUALITY_IMAGES_PER_SPLIT
        )
    ),

    "Validation": (
        raw_val,
        stratified_audit_indices(
            raw_val,
            min(
                QUALITY_IMAGES_PER_SPLIT,
                len(raw_val)
            )
        )
    ),

    "Test": (
        raw_test,
        stratified_audit_indices(
            raw_test,
            min(
                QUALITY_IMAGES_PER_SPLIT,
                len(raw_test)
            )
        )
    )
}


quality_rows = []
corrupted_files = []
hash_to_files = defaultdict(list)


for split_name, (dataset, indices) in audit_sets.items():

    print(
        f"Auditing {split_name}: "
        f"{len(indices):,} images"
    )

    for index in tqdm(
        indices,
        desc=f"{split_name} audit",
        leave=False
    ):

        path, target = dataset.samples[int(index)]

        path_obj = Path(path)

        try:

            with open(path, "rb") as file:
                raw_bytes = file.read()

            exact_hash = hashlib.sha1(
                raw_bytes
            ).hexdigest()

            hash_to_files[exact_hash].append(
                path
            )

            with Image.open(path) as image:

                width, height = image.size

                original_mode = image.mode

                rgb = image.convert("RGB")

                small = rgb.resize((64, 64))

                array = np.asarray(
                    small,
                    dtype=np.float32
                ) / 255.0

            quality_rows.append({

                "split":
                    split_name,

                "class":
                    CLASSES[int(target)],

                "filepath":
                    path,

                "extension":
                    path_obj.suffix.lower(),

                "width":
                    width,

                "height":
                    height,

                "aspect_ratio":
                    width / height,

                "mode":
                    original_mode,

                "brightness":
                    float(array.mean()),

                "red_mean":
                    float(array[:, :, 0].mean()),

                "green_mean":
                    float(array[:, :, 1].mean()),

                "blue_mean":
                    float(array[:, :, 2].mean())
            })

        except Exception as error:

            corrupted_files.append({
                "split":
                    split_name,

                "filepath":
                    path,

                "error":
                    str(error)
            })


quality_df = pd.DataFrame(quality_rows)

quality_df.to_csv(
    DATA_DIR / "sampled_image_quality_statistics.csv",
    index=False
)

pd.DataFrame(corrupted_files).to_csv(
    DATA_DIR / "sampled_corrupted_images.csv",
    index=False
)


duplicates = []

for hash_value, files in hash_to_files.items():

    if len(files) > 1:

        duplicates.append({
            "sha1":
                hash_value,

            "count":
                len(files),

            "files":
                " | ".join(files)
        })


duplicate_df = pd.DataFrame(duplicates)

duplicate_df.to_csv(
    DATA_DIR / "sampled_exact_duplicates.csv",
    index=False
)


print(
    "Sampled corrupted/unreadable images:",
    len(corrupted_files)
)

print(
    "Sampled duplicate-content groups:",
    len(duplicates)
)


# ================================================================================================
# SECTION 9 — FORMAT / MODE SUMMARY
# ================================================================================================

format_summary = (
    quality_df
    .groupby(["split", "extension"])
    .size()
    .reset_index(name="count")
)

mode_summary = (
    quality_df
    .groupby(["split", "mode"])
    .size()
    .reset_index(name="count")
)

format_summary.to_csv(
    DATA_DIR / "image_format_summary.csv",
    index=False
)

mode_summary.to_csv(
    DATA_DIR / "image_mode_summary.csv",
    index=False
)

print("\nIMAGE FORMAT SUMMARY")
print(format_summary.to_string(index=False))

print("\nIMAGE MODE SUMMARY")
print(mode_summary.to_string(index=False))


# ================================================================================================
# SECTION 10 — DIMENSION / ASPECT RATIO VISUALIZATION
# ================================================================================================

fig, ax = plt.subplots(figsize=(9, 5))

for class_name in CLASSES:

    values = quality_df[
        quality_df["class"] == class_name
    ]["width"]

    ax.hist(
        values,
        bins=30,
        alpha=0.55,
        label=class_name.title()
    )

ax.set_xlabel("Original Image Width (pixels)")
ax.set_ylabel("Frequency")
ax.set_title("Original Image-Width Distribution")
ax.legend()

fig.tight_layout()

fig.savefig(
    DATA_DIR / "02_original_width_distribution.png",
    dpi=220
)

plt.show()
plt.close(fig)


fig, ax = plt.subplots(figsize=(9, 5))

for class_name in CLASSES:

    values = quality_df[
        quality_df["class"] == class_name
    ]["aspect_ratio"]

    ax.hist(
        values,
        bins=30,
        alpha=0.55,
        label=class_name.title()
    )

ax.set_xlabel("Width / Height")
ax.set_ylabel("Frequency")
ax.set_title("Original Aspect-Ratio Distribution")
ax.legend()

fig.tight_layout()

fig.savefig(
    DATA_DIR / "03_aspect_ratio_distribution.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 11 — BRIGHTNESS / COLOR STATISTICS
# ================================================================================================

fig, ax = plt.subplots(figsize=(9, 5))

for class_name in CLASSES:

    values = quality_df[
        quality_df["class"] == class_name
    ]["brightness"]

    ax.hist(
        values,
        bins=35,
        alpha=0.55,
        label=class_name.title()
    )

ax.set_xlabel("Mean Normalized Pixel Intensity")
ax.set_ylabel("Frequency")
ax.set_title("Class-Wise Brightness Distribution")
ax.legend()

fig.tight_layout()

fig.savefig(
    DATA_DIR / "04_brightness_distribution.png",
    dpi=220
)

plt.show()
plt.close(fig)


class_statistics = (
    quality_df
    .groupby("class")
    [
        [
            "width",
            "height",
            "aspect_ratio",
            "brightness",
            "red_mean",
            "green_mean",
            "blue_mean"
        ]
    ]
    .agg(["mean", "std", "median"])
)

class_statistics.to_csv(
    DATA_DIR / "class_image_statistics.csv"
)

print("\nCLASS-WISE IMAGE STATISTICS")
print(class_statistics)


# ================================================================================================
# SECTION 12 — PREPROCESSING
# ================================================================================================

print("\n" + "=" * 110)
print("SECTION 12 — STUDENT-DESIGNED PREPROCESSING")
print("=" * 110)

IMAGENET_MEAN = (
    0.485,
    0.456,
    0.406
)

IMAGENET_STD = (
    0.229,
    0.224,
    0.225
)


train_transform = transforms.Compose([

    transforms.Lambda(
        lambda image:
        image.convert("RGB")
    ),

    transforms.RandomResizedCrop(
        IMAGE_SIZE,
        scale=(0.90, 1.00),
        ratio=(0.95, 1.05)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        IMAGENET_MEAN,
        IMAGENET_STD
    )
])


eval_transform = transforms.Compose([

    transforms.Lambda(
        lambda image:
        image.convert("RGB")
    ),

    transforms.Resize(256),

    transforms.CenterCrop(
        IMAGE_SIZE
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        IMAGENET_MEAN,
        IMAGENET_STD
    )
])


print("RGB conversion            : Yes")
print("Target resolution         : 224 × 224")
print("Training random crop      : Yes")
print("Horizontal augmentation   : Yes")
print("ImageNet normalization    : Yes")
print("Validation deterministic  : Yes")
print("Test deterministic        : Yes")
print("Leakage rule              : Test set never used for model selection")


# ================================================================================================
# SECTION 13 — BEFORE / AFTER PREPROCESSING
# ================================================================================================

def denormalize_tensor(tensor):

    mean = torch.tensor(
        IMAGENET_MEAN
    ).view(3, 1, 1)

    std = torch.tensor(
        IMAGENET_STD
    ).view(3, 1, 1)

    tensor = (
        tensor.cpu() * std + mean
    )

    return tensor.clamp(0, 1)


train_targets_array = np.asarray(
    raw_train.targets
)

sample_indices = []

for class_name in ["fake", "real"]:

    class_id = CLASS_TO_ID[class_name]

    class_indices = np.flatnonzero(
        train_targets_array == class_id
    )

    sample_indices.extend(
        class_indices[:3].tolist()
    )


fig, axes = plt.subplots(
    len(sample_indices),
    2,
    figsize=(8, 3 * len(sample_indices))
)

for row, index in enumerate(sample_indices):

    path, target = raw_train.samples[index]

    with Image.open(path) as image:

        original = image.convert("RGB")

        processed = eval_transform(
            original
        )

    axes[row, 0].imshow(original)
    axes[row, 0].set_title(
        f"Original — {CLASSES[target].title()}"
    )
    axes[row, 0].axis("off")

    axes[row, 1].imshow(
        denormalize_tensor(
            processed
        ).permute(1, 2, 0)
    )
    axes[row, 1].set_title(
        "After RGB + Resize/Crop + Normalize"
    )
    axes[row, 1].axis("off")


fig.suptitle(
    "Representative Images Before and After Preprocessing",
    fontsize=15
)

fig.tight_layout()

fig.savefig(
    DATA_DIR / "05_before_after_preprocessing.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 14 — AUGMENTATION VISUALIZATION
# ================================================================================================

example_index = sample_indices[0]

example_path, example_target = (
    raw_train.samples[example_index]
)

with Image.open(example_path) as image:
    example_image = image.convert("RGB")


fig, axes = plt.subplots(
    2,
    4,
    figsize=(12, 6)
)

for axis in axes.flatten():

    augmented = train_transform(
        example_image
    )

    axis.imshow(
        denormalize_tensor(
            augmented
        ).permute(1, 2, 0)
    )

    axis.axis("off")


fig.suptitle(
    "Training-Time Augmentation Examples",
    fontsize=15
)

fig.tight_layout()

fig.savefig(
    DATA_DIR / "06_training_augmentation_examples.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 15 — TRANSFORMED DATASETS
# ================================================================================================

train_dataset = datasets.ImageFolder(
    TRAIN_PATH,
    transform=train_transform
)

val_dataset = datasets.ImageFolder(
    VAL_PATH,
    transform=eval_transform
)

test_dataset = datasets.ImageFolder(
    TEST_PATH,
    transform=eval_transform
)


# ================================================================================================
# SECTION 16 — IDENTICAL STRATIFIED BAKE-OFF SUBSETS
# ================================================================================================

def make_stratified_indices(
    targets,
    requested_size
):

    targets = np.asarray(
        targets,
        dtype=np.int64
    )

    all_indices = np.arange(
        len(targets)
    )

    selected, _ = train_test_split(
        all_indices,
        train_size=requested_size,
        random_state=SEED,
        stratify=targets
    )

    return selected.tolist()


bake_train_indices = (
    make_stratified_indices(
        raw_train.targets,
        BAKE_TRAIN_SIZE
    )
)

bake_val_indices = (
    make_stratified_indices(
        raw_val.targets,
        BAKE_VAL_SIZE
    )
)


bake_train_dataset = Subset(
    train_dataset,
    bake_train_indices
)

bake_val_dataset = Subset(
    val_dataset,
    bake_val_indices
)


print("\n" + "=" * 110)
print("SECTION 16 — CONTROLLED CNN BAKE-OFF")
print("=" * 110)

print(
    "Training subset:",
    f"{len(bake_train_dataset):,}"
)

print(
    "Validation subset:",
    f"{len(bake_val_dataset):,}"
)

print(
    "Epochs/model:",
    BAKE_EPOCHS
)

print(
    "✅ Same seed/subset method as transformer experiment."
)


# ================================================================================================
# SECTION 17 — DATALOADERS
# ================================================================================================

def seed_worker(worker_id):

    worker_seed = SEED + worker_id

    np.random.seed(worker_seed)
    random.seed(worker_seed)


def make_loader(
    dataset,
    batch_size,
    shuffle
):

    generator = torch.Generator()

    generator.manual_seed(SEED)

    return DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle,
        num_workers=NUM_WORKERS,
        pin_memory=True,
        persistent_workers=NUM_WORKERS > 0,
        worker_init_fn=seed_worker,
        generator=generator
    )


# ================================================================================================
# SECTION 18 — SELECT CNN CHECKPOINTS
# ================================================================================================

print("\n" + "=" * 110)
print("SECTION 18 — CNN MODEL SELECTION")
print("=" * 110)

available_models = set(
    timm.list_models(
        pretrained=True
    )
)


def choose_model(
    preferred,
    fallbacks,
    search_pattern
):

    candidates = [
        preferred
    ] + fallbacks

    for name in candidates:

        if name in available_models:
            return name

    matches = timm.list_models(
        search_pattern,
        pretrained=True
    )

    if not matches:
        raise RuntimeError(
            f"No pretrained model available for {search_pattern}"
        )

    return matches[0]


RESNET_MODEL_NAME = choose_model(
    preferred="resnet50.a1_in1k",
    fallbacks=[
        "resnet50.a2_in1k",
        "resnet50.tv_in1k",
        "resnet50"
    ],
    search_pattern="resnet50*"
)


EFFNET_MODEL_NAME = choose_model(
    preferred="efficientnet_b0.ra_in1k",
    fallbacks=[
        "efficientnet_b0.rwightman_in1k",
        "efficientnet_b0"
    ],
    search_pattern="efficientnet_b0*"
)


print("✅ ResNet50:", RESNET_MODEL_NAME)
print("✅ EfficientNet-B0:", EFFNET_MODEL_NAME)


MODEL_NAMES = {
    "ResNet50":
        RESNET_MODEL_NAME,

    "EfficientNet-B0":
        EFFNET_MODEL_NAME
}


# ================================================================================================
# SECTION 19 — MODEL CREATION / COMPLEXITY
# ================================================================================================

def create_model(model_name):

    return timm.create_model(
        model_name,
        pretrained=True,
        num_classes=2
    )


complexity_rows = [
    {
        "model":
            "ViT-Base/16",

        "family":
            "Transformer",

        "checkpoint":
            "vit_base_patch16_224.augreg2_in21k_ft_in1k",

        "parameters_millions":
            85.80
    },

    {
        "model":
            "Swin-Tiny",

        "family":
            "Transformer",

        "checkpoint":
            "swin_tiny_patch4_window7_224.ms_in1k",

        "parameters_millions":
            27.52
    }
]


for architecture in [
    "ResNet50",
    "EfficientNet-B0"
]:

    temp_model = create_model(
        MODEL_NAMES[architecture]
    )

    parameters = sum(
        p.numel()
        for p in temp_model.parameters()
    )

    complexity_rows.append({
        "model":
            architecture,

        "family":
            "CNN",

        "checkpoint":
            MODEL_NAMES[architecture],

        "parameters_millions":
            parameters / 1_000_000
    })

    print(
        architecture,
        f"{parameters/1e6:.2f}M parameters"
    )

    del temp_model

    gc.collect()


complexity_df = pd.DataFrame(
    complexity_rows
)

complexity_df.to_csv(
    COMPARE_DIR / "model_complexity.csv",
    index=False
)


fig, ax = plt.subplots(
    figsize=(10, 5)
)

bars = ax.bar(
    complexity_df["model"],
    complexity_df[
        "parameters_millions"
    ]
)

ax.set_ylabel(
    "Parameters (Millions)"
)

ax.set_title(
    "Parameter Count Across Four Architectures"
)

for bar, value in zip(
    bars,
    complexity_df["parameters_millions"]
):

    ax.text(
        bar.get_x() + bar.get_width()/2,
        value,
        f"{value:.1f}M",
        ha="center",
        va="bottom"
    )

fig.tight_layout()

fig.savefig(
    COMPARE_DIR / "01_parameter_comparison.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 20 — METRIC FUNCTIONS
# ================================================================================================

def numpy_softmax(logits):

    logits = np.asarray(
        logits,
        dtype=np.float64
    )

    logits = (
        logits
        -
        np.max(
            logits,
            axis=1,
            keepdims=True
        )
    )

    exp_logits = np.exp(logits)

    return (
        exp_logits
        /
        exp_logits.sum(
            axis=1,
            keepdims=True
        )
    )


def calculate_metrics(
    logits,
    labels
):

    labels = np.asarray(
        labels,
        dtype=np.int64
    )

    probabilities = numpy_softmax(
        logits
    )

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    fake_probability = probabilities[
        :,
        FAKE_ID
    ]

    binary_true = (
        labels == FAKE_ID
    ).astype(np.int64)

    binary_prediction = (
        predictions == FAKE_ID
    ).astype(np.int64)

    cm = confusion_matrix(
        binary_true,
        binary_prediction,
        labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0.0
    )

    brier = float(
        np.mean(
            (
                fake_probability -
                binary_true
            ) ** 2
        )
    )

    metrics = {

        "accuracy":
            float(
                accuracy_score(
                    labels,
                    predictions
                )
            ),

        "balanced_accuracy":
            float(
                balanced_accuracy_score(
                    labels,
                    predictions
                )
            ),

        "precision_fake":
            float(
                precision_score(
                    binary_true,
                    binary_prediction,
                    zero_division=0
                )
            ),

        "recall_fake":
            float(
                recall_score(
                    binary_true,
                    binary_prediction,
                    zero_division=0
                )
            ),

        "specificity_real":
            float(specificity),

        "f1_fake":
            float(
                f1_score(
                    binary_true,
                    binary_prediction,
                    zero_division=0
                )
            ),

        "auc_fake":
            float(
                roc_auc_score(
                    binary_true,
                    fake_probability
                )
            ),

        "average_precision_fake":
            float(
                average_precision_score(
                    binary_true,
                    fake_probability
                )
            ),

        "kappa":
            float(
                cohen_kappa_score(
                    labels,
                    predictions
                )
            ),

        "mcc":
            float(
                matthews_corrcoef(
                    binary_true,
                    binary_prediction
                )
            ),

        "brier_fake":
            brier
    }

    return {

        "metrics":
            metrics,

        "labels":
            labels,

        "predictions":
            predictions,

        "probabilities":
            probabilities,

        "fake_probability":
            fake_probability,

        "binary_true":
            binary_true,

        "binary_prediction":
            binary_prediction,

        "confusion_matrix":
            cm
    }


# ================================================================================================
# SECTION 21 — EVALUATION
# ================================================================================================

@torch.inference_mode()
def evaluate_model(
    model,
    loader,
    criterion,
    description
):

    model.eval()

    total_loss = 0
    total_samples = 0

    all_logits = []
    all_labels = []

    for images, labels in tqdm(
        loader,
        desc=description,
        leave=False
    ):

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        labels = labels.to(
            DEVICE,
            non_blocking=True
        )

        with torch.amp.autocast(
            device_type="cuda",
            dtype=torch.float16
        ):

            logits = model(images)

            loss = criterion(
                logits,
                labels
            )

        batch_size = images.size(0)

        total_loss += (
            loss.item() * batch_size
        )

        total_samples += batch_size

        all_logits.append(
            logits
            .float()
            .cpu()
            .numpy()
        )

        all_labels.append(
            labels
            .cpu()
            .numpy()
        )


    all_logits = np.concatenate(
        all_logits
    )

    all_labels = np.concatenate(
        all_labels
    )

    result = calculate_metrics(
        all_logits,
        all_labels
    )

    result["loss"] = (
        total_loss /
        total_samples
    )

    result["logits"] = all_logits

    return result


# ================================================================================================
# SECTION 22 — SAFE CHECKPOINT LOADING
# ================================================================================================

def safe_load(path):

    try:

        return torch.load(
            path,
            map_location="cpu",
            weights_only=True
        )

    except TypeError:

        return torch.load(
            path,
            map_location="cpu"
        )


# ================================================================================================
# SECTION 23 — CNN TRAINING FUNCTION
# ================================================================================================

def train_architecture(
    architecture,
    model_name,
    training_data,
    validation_data,
    epochs,
    output_directory,
    keep_model=False
):

    print("\n" + "=" * 110)
    print("TRAINING", architecture)
    print("=" * 110)

    print("Checkpoint:", model_name)
    print("Training images:", f"{len(training_data):,}")
    print("Validation images:", f"{len(validation_data):,}")
    print("Epochs:", epochs)

    gc.collect()
    torch.cuda.empty_cache()

    model = create_model(
        model_name
    ).to(DEVICE)

    criterion = nn.CrossEntropyLoss()

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY
    )

    train_loader = make_loader(
        training_data,
        TRAIN_BATCH_SIZE,
        shuffle=True
    )

    val_loader = make_loader(
        validation_data,
        EVAL_BATCH_SIZE,
        shuffle=False
    )

    scaler = torch.amp.GradScaler(
        "cuda"
    )

    history = {
        "step": [],
        "step_loss": [],
        "epoch": [],
        "train_loss": [],
        "validation_loss": [],
        "validation_accuracy": [],
        "validation_f1": [],
        "validation_auc": []
    }

    best_auc = -np.inf

    best_path = (
        output_directory /
        "best_model_state_dict.pt"
    )

    global_step = 0

    start_time = time.time()

    for epoch in range(
        1,
        epochs + 1
    ):

        print(
            f"\n{architecture} "
            f"EPOCH {epoch}/{epochs}"
        )

        model.train()

        total_train_loss = 0
        total_train_samples = 0

        progress = tqdm(
            train_loader,
            desc=f"{architecture} Epoch {epoch}/{epochs}"
        )

        for images, labels in progress:

            images = images.to(
                DEVICE,
                non_blocking=True
            )

            labels = labels.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            with torch.amp.autocast(
                device_type="cuda",
                dtype=torch.float16
            ):

                logits = model(images)

                loss = criterion(
                    logits,
                    labels
                )

            scaler.scale(
                loss
            ).backward()

            scaler.unscale_(
                optimizer
            )

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                GRAD_CLIP_NORM
            )

            scaler.step(
                optimizer
            )

            scaler.update()

            batch_size = images.size(0)

            total_train_loss += (
                loss.item() *
                batch_size
            )

            total_train_samples += (
                batch_size
            )

            global_step += 1

            history["step"].append(
                global_step
            )

            history["step_loss"].append(
                float(loss.item())
            )

            progress.set_postfix(
                loss=f"{loss.item():.4f}"
            )


        train_loss = (
            total_train_loss /
            total_train_samples
        )

        validation = evaluate_model(
            model,
            val_loader,
            criterion,
            f"{architecture} Validation"
        )

        metrics = validation[
            "metrics"
        ]

        history["epoch"].append(epoch)
        history["train_loss"].append(train_loss)
        history["validation_loss"].append(validation["loss"])
        history["validation_accuracy"].append(metrics["accuracy"])
        history["validation_f1"].append(metrics["f1_fake"])
        history["validation_auc"].append(metrics["auc_fake"])

        print(
            f"\nTrain loss        : "
            f"{train_loss:.6f}"
        )

        print(
            f"Validation loss   : "
            f"{validation['loss']:.6f}"
        )

        print(
            f"Accuracy          : "
            f"{metrics['accuracy']:.6f}"
        )

        print(
            f"Balanced Accuracy : "
            f"{metrics['balanced_accuracy']:.6f}"
        )

        print(
            f"Fake Precision    : "
            f"{metrics['precision_fake']:.6f}"
        )

        print(
            f"Fake Recall       : "
            f"{metrics['recall_fake']:.6f}"
        )

        print(
            f"Real Specificity  : "
            f"{metrics['specificity_real']:.6f}"
        )

        print(
            f"Fake F1           : "
            f"{metrics['f1_fake']:.6f}"
        )

        print(
            f"Fake ROC-AUC      : "
            f"{metrics['auc_fake']:.6f}"
        )

        print(
            f"Fake PR-AUC       : "
            f"{metrics['average_precision_fake']:.6f}"
        )

        print(
            f"Kappa              : "
            f"{metrics['kappa']:.6f}"
        )

        print(
            f"MCC                : "
            f"{metrics['mcc']:.6f}"
        )

        print(
            f"Brier Score        : "
            f"{metrics['brier_fake']:.6f}"
        )


        if metrics[
            "auc_fake"
        ] > best_auc:

            best_auc = metrics[
                "auc_fake"
            ]

            torch.save(
                model.state_dict(),
                best_path
            )

            print(
                "✅ Best checkpoint saved."
            )


    training_seconds = (
        time.time() - start_time
    )

    model.load_state_dict(
        safe_load(best_path),
        strict=True
    )

    final_validation = evaluate_model(
        model,
        val_loader,
        criterion,
        f"{architecture} Best Validation"
    )

    metrics = final_validation[
        "metrics"
    ]

    pd.DataFrame({
        "step":
            history["step"],

        "loss":
            history["step_loss"]
    }).to_csv(
        output_directory /
        "training_step_loss.csv",
        index=False
    )

    pd.DataFrame({
        "epoch":
            history["epoch"],

        "train_loss":
            history["train_loss"],

        "validation_loss":
            history["validation_loss"],

        "validation_accuracy":
            history["validation_accuracy"],

        "validation_f1_fake":
            history["validation_f1"],

        "validation_auc_fake":
            history["validation_auc"]
    }).to_csv(
        output_directory /
        "epoch_history.csv",
        index=False
    )

    with open(
        output_directory /
        "results.json",
        "w"
    ) as file:

        json.dump(
            {
                "architecture":
                    architecture,

                "checkpoint":
                    model_name,

                "training_seconds":
                    training_seconds,

                "metrics":
                    metrics
            },
            file,
            indent=2
        )


    result = {

        "architecture":
            architecture,

        "model_name":
            model_name,

        "model_path":
            str(best_path),

        "metrics":
            metrics,

        "validation":
            final_validation,

        "training_seconds":
            training_seconds,

        "history":
            history
    }


    if keep_model:

        result["model"] = model

    else:

        model = model.cpu()

        del model

        gc.collect()

        torch.cuda.empty_cache()


    print(
        f"\n✅ {architecture} complete "
        f"in {training_seconds/60:.2f} minutes."
    )

    return result


# ================================================================================================
# SECTION 24 — VISUALIZATION FUNCTIONS
# ================================================================================================

def plot_loss(result, output_path):

    losses = np.asarray(
        result["history"]["step_loss"]
    )

    steps = np.asarray(
        result["history"]["step"]
    )

    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    if len(losses) >= 20:

        window = min(
            25,
            len(losses)
        )

        smoothed = np.convolve(
            losses,
            np.ones(window) / window,
            mode="valid"
        )

        ax.plot(
            steps[window-1:],
            smoothed
        )

    else:

        ax.plot(
            steps,
            losses
        )

    ax.set_xlabel(
        "Optimizer Step"
    )

    ax.set_ylabel(
        "Cross-Entropy Loss"
    )

    ax.set_title(
        f"{result['architecture']} Training Loss"
    )

    fig.tight_layout()

    fig.savefig(
        output_path,
        dpi=220
    )

    plt.show()
    plt.close(fig)


def plot_cm(
    cm,
    title,
    output_path,
    normalized=False
):

    display = np.asarray(
        cm,
        dtype=np.float64
    )

    if normalized:

        denominator = display.sum(
            axis=1,
            keepdims=True
        )

        denominator[
            denominator == 0
        ] = 1

        display = (
            display /
            denominator
        )


    fig, ax = plt.subplots(
        figsize=(6, 5)
    )

    image = ax.imshow(
        display
    )

    ax.set_xticks([0, 1])
    ax.set_yticks([0, 1])

    ax.set_xticklabels(
        ["Real", "Fake"]
    )

    ax.set_yticklabels(
        ["Real", "Fake"]
    )

    ax.set_xlabel(
        "Predicted Class"
    )

    ax.set_ylabel(
        "Actual Class"
    )

    ax.set_title(title)


    for row in range(2):

        for column in range(2):

            if normalized:

                text = (
                    f"{display[row,column]:.2%}"
                )

            else:

                text = (
                    f"{int(cm[row,column]):,}"
                )

            ax.text(
                column,
                row,
                text,
                ha="center",
                va="center"
            )


    fig.colorbar(
        image,
        ax=ax
    )

    fig.tight_layout()

    fig.savefig(
        output_path,
        dpi=220
    )

    plt.show()
    plt.close(fig)


def plot_probability(
    evaluation,
    title,
    output_path
):

    probability = evaluation[
        "fake_probability"
    ]

    true = evaluation[
        "binary_true"
    ]

    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    ax.hist(
        probability[true == 1],
        bins=40,
        alpha=0.6,
        label="Actual Fake"
    )

    ax.hist(
        probability[true == 0],
        bins=40,
        alpha=0.6,
        label="Actual Real"
    )

    ax.set_xlabel(
        "Predicted Probability of Fake"
    )

    ax.set_ylabel(
        "Images"
    )

    ax.set_title(title)

    ax.legend()

    fig.tight_layout()

    fig.savefig(
        output_path,
        dpi=220
    )

    plt.show()
    plt.close(fig)


# ================================================================================================
# SECTION 25 — TRAIN RESNET50
# ================================================================================================

print("\n" + "#" * 110)
print("SECTION 25 — RESNET50 BAKE-OFF")
print("#" * 110)

resnet_result = train_architecture(
    architecture="ResNet50",
    model_name=RESNET_MODEL_NAME,
    training_data=bake_train_dataset,
    validation_data=bake_val_dataset,
    epochs=BAKE_EPOCHS,
    output_directory=RESNET_DIR
)

plot_loss(
    resnet_result,
    RESNET_DIR /
    "training_loss.png"
)

plot_cm(
    resnet_result[
        "validation"
    ]["confusion_matrix"],
    "ResNet50 Validation Confusion Matrix",
    RESNET_DIR /
    "confusion_matrix.png"
)

plot_cm(
    resnet_result[
        "validation"
    ]["confusion_matrix"],
    "ResNet50 Normalized Validation Confusion Matrix",
    RESNET_DIR /
    "confusion_matrix_normalized.png",
    normalized=True
)

plot_probability(
    resnet_result[
        "validation"
    ],
    "ResNet50 Validation Probability Distribution",
    RESNET_DIR /
    "probability_distribution.png"
)


# ================================================================================================
# SECTION 26 — TRAIN EFFICIENTNET-B0
# ================================================================================================

print("\n" + "#" * 110)
print("SECTION 26 — EFFICIENTNET-B0 BAKE-OFF")
print("#" * 110)

effnet_result = train_architecture(
    architecture="EfficientNet-B0",
    model_name=EFFNET_MODEL_NAME,
    training_data=bake_train_dataset,
    validation_data=bake_val_dataset,
    epochs=BAKE_EPOCHS,
    output_directory=EFFNET_DIR
)

plot_loss(
    effnet_result,
    EFFNET_DIR /
    "training_loss.png"
)

plot_cm(
    effnet_result[
        "validation"
    ]["confusion_matrix"],
    "EfficientNet-B0 Validation Confusion Matrix",
    EFFNET_DIR /
    "confusion_matrix.png"
)

plot_cm(
    effnet_result[
        "validation"
    ]["confusion_matrix"],
    "EfficientNet-B0 Normalized Validation Confusion Matrix",
    EFFNET_DIR /
    "confusion_matrix_normalized.png",
    normalized=True
)

plot_probability(
    effnet_result[
        "validation"
    ],
    "EfficientNet-B0 Validation Probability Distribution",
    EFFNET_DIR /
    "probability_distribution.png"
)


# ================================================================================================
# SECTION 27 — REUSE CONFIRMED TRANSFORMER RESULTS
# ================================================================================================

print("\n" + "=" * 110)
print("SECTION 27 — IMPORTING PREVIOUS TRANSFORMER RESULTS")
print("=" * 110)

print(
    "⚠️ ViT and Swin are NOT being retrained.\n"
    "The following values come from the previously completed "
    "controlled transformer experiment."
)


VIT_PREVIOUS = {

    "model":
        "ViT-Base/16",

    "family":
        "Transformer",

    "checkpoint":
        "vit_base_patch16_224.augreg2_in21k_ft_in1k",

    "accuracy":
        0.955500,

    "balanced_accuracy":
        0.955500,

    "precision_fake":
        0.977964,

    "recall_fake":
        0.932000,

    "specificity_real":
        0.979000,

    "f1_fake":
        0.954429,

    "auc_fake":
        0.992458,

    "average_precision_fake":
        0.992449,

    "kappa":
        0.911000,

    "mcc":
        0.912008,

    "brier_fake":
        0.035432,

    "training_seconds":
        118.007673,

    "parameters_millions":
        85.80
}


SWIN_PREVIOUS = {

    "model":
        "Swin-Tiny",

    "family":
        "Transformer",

    "checkpoint":
        "swin_tiny_patch4_window7_224.ms_in1k",

    "accuracy":
        0.969500,

    "balanced_accuracy":
        0.969500,

    "precision_fake":
        0.958944,

    "recall_fake":
        0.981000,

    "specificity_real":
        0.958000,

    "f1_fake":
        0.969847,

    "auc_fake":
        0.997089,

    "average_precision_fake":
        0.997183,

    "kappa":
        0.939000,

    "mcc":
        0.939248,

    "brier_fake":
        0.021871,

    "training_seconds":
        95.108889,

    "parameters_millions":
        27.52
}


print(
    "ViT ROC-AUC:",
    VIT_PREVIOUS["auc_fake"]
)

print(
    "Swin ROC-AUC:",
    SWIN_PREVIOUS["auc_fake"]
)


# ================================================================================================
# SECTION 28 — BUILD FOUR-MODEL COMPARISON TABLE
# ================================================================================================

complexity_lookup = {
    row["model"]:
    row["parameters_millions"]
    for row in complexity_rows
}


def cnn_comparison_row(
    result,
    model_display
):

    return {

        "model":
            model_display,

        "family":
            "CNN",

        "checkpoint":
            result["model_name"],

        **result["metrics"],

        "training_seconds":
            result["training_seconds"],

        "parameters_millions":
            complexity_lookup[
                model_display
            ]
    }


comparison_rows = [
    VIT_PREVIOUS,
    SWIN_PREVIOUS,
    cnn_comparison_row(
        resnet_result,
        "ResNet50"
    ),
    cnn_comparison_row(
        effnet_result,
        "EfficientNet-B0"
    )
]


comparison_df = pd.DataFrame(
    comparison_rows
)

comparison_df = comparison_df.sort_values(
    "auc_fake",
    ascending=False
).reset_index(drop=True)


comparison_df.to_csv(
    COMPARE_DIR /
    "FOUR_MODEL_BAKEOFF_RESULTS.csv",
    index=False
)


print("\n" + "=" * 150)
print("FOUR-MODEL CONTROLLED BAKE-OFF RESULTS")
print("=" * 150)

print(
    comparison_df[
        [
            "model",
            "family",
            "accuracy",
            "balanced_accuracy",
            "precision_fake",
            "recall_fake",
            "specificity_real",
            "f1_fake",
            "auc_fake",
            "average_precision_fake",
            "kappa",
            "mcc",
            "brier_fake",
            "parameters_millions",
            "training_seconds"
        ]
    ].to_string(index=False)
)


# ================================================================================================
# SECTION 29 — FOUR-MODEL PERFORMANCE PLOT
# ================================================================================================

metrics_for_plot = [
    "accuracy",
    "balanced_accuracy",
    "f1_fake",
    "auc_fake",
    "average_precision_fake",
    "mcc"
]

labels_for_plot = [
    "Accuracy",
    "Balanced Acc.",
    "Fake F1",
    "ROC-AUC",
    "PR-AUC",
    "MCC"
]


fig, ax = plt.subplots(
    figsize=(15, 7)
)

x = np.arange(
    len(metrics_for_plot)
)

width = 0.19

models_order = [
    "ViT-Base/16",
    "Swin-Tiny",
    "ResNet50",
    "EfficientNet-B0"
]


for model_number, model_name in enumerate(
    models_order
):

    row = comparison_df[
        comparison_df["model"] ==
        model_name
    ].iloc[0]

    values = [
        float(row[m])
        for m in metrics_for_plot
    ]

    positions = (
        x +
        (
            model_number -
            1.5
        ) * width
    )

    ax.bar(
        positions,
        values,
        width,
        label=model_name
    )


ax.set_xticks(x)
ax.set_xticklabels(
    labels_for_plot
)

ax.set_ylim(
    0,
    1.06
)

ax.set_ylabel(
    "Score"
)

ax.set_title(
    "CNN vs Transformer — Controlled Validation Performance"
)

ax.legend(
    ncol=2
)

fig.tight_layout()

fig.savefig(
    COMPARE_DIR /
    "02_four_model_metric_comparison.png",
    dpi=250
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 30 — TRAINING TIME
# ================================================================================================

ordered_df = (
    comparison_df
    .set_index("model")
    .loc[models_order]
    .reset_index()
)


fig, ax = plt.subplots(
    figsize=(10, 5)
)

values = (
    ordered_df[
        "training_seconds"
    ] / 60
)

bars = ax.bar(
    ordered_df["model"],
    values
)

ax.set_ylabel(
    "Training Time (minutes)"
)

ax.set_title(
    "One-Epoch Bake-Off Training Time"
)

for bar, value in zip(
    bars,
    values
):

    ax.text(
        bar.get_x() +
        bar.get_width()/2,
        value,
        f"{value:.2f}",
        ha="center",
        va="bottom"
    )

fig.tight_layout()

fig.savefig(
    COMPARE_DIR /
    "03_training_time_comparison.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 31 — ACCURACY VS MODEL SIZE
# ================================================================================================

fig, ax = plt.subplots(
    figsize=(8, 6)
)

for _, row in comparison_df.iterrows():

    ax.scatter(
        row["parameters_millions"],
        row["accuracy"],
        s=100
    )

    ax.annotate(
        row["model"],
        (
            row["parameters_millions"],
            row["accuracy"]
        ),
        xytext=(5, 5),
        textcoords="offset points"
    )


ax.set_xlabel(
    "Parameters (Millions)"
)

ax.set_ylabel(
    "Validation Accuracy"
)

ax.set_title(
    "Predictive Performance vs Model Complexity"
)

fig.tight_layout()

fig.savefig(
    COMPARE_DIR /
    "04_accuracy_vs_model_size.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 32 — CNN ROC CURVES + PREVIOUS TRANSFORMER AUC REFERENCE
# ================================================================================================

fig, ax = plt.subplots(
    figsize=(8, 6)
)

for result in [
    resnet_result,
    effnet_result
]:

    evaluation = result[
        "validation"
    ]

    fpr, tpr, _ = roc_curve(
        evaluation["binary_true"],
        evaluation["fake_probability"]
    )

    ax.plot(
        fpr,
        tpr,
        linewidth=2,
        label=(
            f"{result['architecture']} "
            f"(AUC={result['metrics']['auc_fake']:.4f})"
        )
    )


ax.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random"
)

ax.set_xlabel(
    "False Positive Rate"
)

ax.set_ylabel(
    "True Positive Rate"
)

ax.set_title(
    "CNN Validation ROC Curves"
)

ax.legend(
    loc="lower right"
)

fig.tight_layout()

fig.savefig(
    COMPARE_DIR /
    "05_cnn_roc_curves.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 33 — CNN PRECISION-RECALL CURVES
# ================================================================================================

fig, ax = plt.subplots(
    figsize=(8, 6)
)

for result in [
    resnet_result,
    effnet_result
]:

    evaluation = result[
        "validation"
    ]

    precision_curve, recall_curve, _ = (
        precision_recall_curve(
            evaluation["binary_true"],
            evaluation["fake_probability"]
        )
    )

    ax.plot(
        recall_curve,
        precision_curve,
        linewidth=2,
        label=(
            f"{result['architecture']} "
            f"(AP="
            f"{result['metrics']['average_precision_fake']:.4f})"
        )
    )


ax.set_xlabel(
    "Recall"
)

ax.set_ylabel(
    "Precision"
)

ax.set_title(
    "CNN Validation Precision-Recall Curves"
)

ax.legend()

fig.tight_layout()

fig.savefig(
    COMPARE_DIR /
    "06_cnn_precision_recall_curves.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 34 — CNN VS TRANSFORMER FAMILY SUMMARY
# ================================================================================================

family_summary = (
    comparison_df
    .groupby("family")
    [
        [
            "accuracy",
            "f1_fake",
            "auc_fake",
            "average_precision_fake",
            "mcc",
            "parameters_millions",
            "training_seconds"
        ]
    ]
    .mean()
    .reset_index()
)


family_summary.to_csv(
    COMPARE_DIR /
    "architecture_family_summary.csv",
    index=False
)


print("\nARCHITECTURE FAMILY SUMMARY")
print(
    family_summary.to_string(
        index=False
    )
)


# ================================================================================================
# SECTION 35 — WINNER SELECTION
# ================================================================================================

winner_row = comparison_df.iloc[0]

WINNER = str(
    winner_row["model"]
)

WINNER_FAMILY = str(
    winner_row["family"]
)

WINNER_AUC = float(
    winner_row["auc_fake"]
)


print("\n" + "=" * 110)
print("SECTION 35 — FOUR-MODEL WINNER")
print("=" * 110)

print("🏆 Winner:", WINNER)
print("Family:", WINNER_FAMILY)
print(
    "Validation fake ROC-AUC:",
    f"{WINNER_AUC:.6f}"
)

print(
    "Selection criterion: "
    "Validation fake-class ROC-AUC"
)


# ================================================================================================
# SECTION 36 — GRAD-CAM IMPLEMENTATION FOR CNNs
# ================================================================================================

print("\n" + "=" * 110)
print("SECTION 36 — CNN EXPLAINABILITY WITH GRAD-CAM")
print("=" * 110)


class GradCAM:

    def __init__(
        self,
        model,
        target_layer
    ):

        self.model = model
        self.activations = None
        self.gradients = None

        self.forward_hook = (
            target_layer.register_forward_hook(
                self._save_activation
            )
        )

        self.backward_hook = (
            target_layer.register_full_backward_hook(
                self._save_gradient
            )
        )


    def _save_activation(
        self,
        module,
        inputs,
        output
    ):

        self.activations = (
            output.detach()
        )


    def _save_gradient(
        self,
        module,
        grad_input,
        grad_output
    ):

        self.gradients = (
            grad_output[0].detach()
        )


    def generate(
        self,
        input_tensor,
        class_id
    ):

        self.model.zero_grad(
            set_to_none=True
        )

        logits = self.model(
            input_tensor
        )

        score = logits[
            0,
            class_id
        ]

        score.backward()

        gradients = self.gradients
        activations = self.activations

        weights = gradients.mean(
            dim=(2, 3),
            keepdim=True
        )

        cam = (
            weights *
            activations
        ).sum(
            dim=1,
            keepdim=True
        )

        cam = F.relu(cam)

        cam = F.interpolate(
            cam,
            size=(
                IMAGE_SIZE,
                IMAGE_SIZE
            ),
            mode="bilinear",
            align_corners=False
        )

        cam = cam[
            0,
            0
        ].cpu().numpy()

        cam -= cam.min()

        if cam.max() > 0:
            cam /= cam.max()

        probability = torch.softmax(
            logits.detach(),
            dim=1
        )[0].cpu().numpy()

        return cam, probability


    def close(self):

        self.forward_hook.remove()
        self.backward_hook.remove()


def gradcam_model(
    architecture
):

    if architecture == "ResNet50":

        model_name = RESNET_MODEL_NAME

        state_path = (
            RESNET_DIR /
            "best_model_state_dict.pt"
        )

    else:

        model_name = EFFNET_MODEL_NAME

        state_path = (
            EFFNET_DIR /
            "best_model_state_dict.pt"
        )


    model = create_model(
        model_name
    )

    model.load_state_dict(
        safe_load(state_path),
        strict=True
    )

    model = model.to(
        DEVICE
    )

    model.eval()

    return model


def create_gradcam_figure(
    architecture,
    number_images=6
):

    model = gradcam_model(
        architecture
    )

    if architecture == "ResNet50":

        target_layer = (
            model.layer4[-1]
        )

    else:

        target_layer = (
            model.conv_head
        )


    cam_engine = GradCAM(
        model,
        target_layer
    )


    selected = []

    val_targets = np.asarray(
        raw_val.targets
    )


    for class_id in [
        FAKE_ID,
        REAL_ID
    ]:

        class_positions = np.flatnonzero(
            val_targets == class_id
        )[:number_images // 2]

        selected.extend(
            class_positions.tolist()
        )


    fig, axes = plt.subplots(
        len(selected),
        3,
        figsize=(
            10,
            3.1 * len(selected)
        )
    )


    for row, index in enumerate(
        selected
    ):

        path, true_id = (
            raw_val.samples[
                int(index)
            ]
        )

        with Image.open(path) as image:

            rgb = image.convert(
                "RGB"
            )

            tensor = eval_transform(
                rgb
            )

        input_tensor = (
            tensor.unsqueeze(0)
            .to(DEVICE)
        )

        with torch.no_grad():

            logits = model(
                input_tensor
            )

            pred_id = int(
                logits.argmax(
                    dim=1
                ).item()
            )


        cam, probabilities = (
            cam_engine.generate(
                input_tensor,
                pred_id
            )
        )


        display_image = (
            denormalize_tensor(
                tensor
            )
            .permute(1, 2, 0)
            .numpy()
        )


        axes[row, 0].imshow(
            display_image
        )

        axes[row, 0].set_title(
            f"Original\n"
            f"True: {CLASSES[true_id]}"
        )

        axes[row, 0].axis("off")


        axes[row, 1].imshow(
            cam
        )

        axes[row, 1].set_title(
            "Grad-CAM"
        )

        axes[row, 1].axis("off")


        axes[row, 2].imshow(
            display_image
        )

        axes[row, 2].imshow(
            cam,
            alpha=0.45,
            cmap="jet"
        )

        axes[row, 2].set_title(
            f"Prediction: "
            f"{CLASSES[pred_id]}\n"
            f"P(fake)="
            f"{probabilities[FAKE_ID]:.3f}"
        )

        axes[row, 2].axis("off")


    fig.suptitle(
        f"{architecture} — Grad-CAM Explanation",
        fontsize=15
    )

    fig.tight_layout()

    output_path = (
        EXPLAIN_DIR /
        f"{architecture.replace('-', '_')}_gradcam.png"
    )

    fig.savefig(
        output_path,
        dpi=220
    )

    plt.show()
    plt.close(fig)


    cam_engine.close()

    model = model.cpu()

    del model

    gc.collect()

    torch.cuda.empty_cache()


create_gradcam_figure(
    "ResNet50"
)

create_gradcam_figure(
    "EfficientNet-B0"
)


# ================================================================================================
# SECTION 37 — MISCLASSIFICATION VISUALIZATION
# ================================================================================================

def plot_errors(
    result,
    name,
    output_path,
    maximum=12
):

    evaluation = result[
        "validation"
    ]

    predictions = evaluation[
        "predictions"
    ]

    labels = evaluation[
        "labels"
    ]

    fake_probability = evaluation[
        "fake_probability"
    ]

    incorrect = np.where(
        predictions != labels
    )[0]


    if len(incorrect) == 0:

        print(
            name,
            "has no validation errors."
        )

        return


    confidence = np.where(
        predictions == FAKE_ID,
        fake_probability,
        1 - fake_probability
    )

    incorrect = sorted(
        incorrect,
        key=lambda i:
            confidence[i],
        reverse=True
    )[:maximum]


    rows = math.ceil(
        len(incorrect) / 4
    )

    fig, axes = plt.subplots(
        rows,
        4,
        figsize=(
            14,
            rows * 3.4
        ),
        squeeze=False
    )

    axes = axes.flatten()

    for axis in axes:
        axis.axis("off")


    for plot_number, position in enumerate(
        incorrect
    ):

        original_index = (
            bake_val_indices[
                position
            ]
        )

        path, _ = (
            raw_val.samples[
                original_index
            ]
        )

        with Image.open(path) as image:

            image = image.convert(
                "RGB"
            )

            axes[
                plot_number
            ].imshow(
                image
            )


        true_name = CLASSES[
            int(labels[position])
        ]

        pred_name = CLASSES[
            int(predictions[position])
        ]

        axes[
            plot_number
        ].set_title(
            f"True: {true_name}\n"
            f"Pred: {pred_name}\n"
            f"P(fake): "
            f"{fake_probability[position]:.3f}"
        )

        axes[
            plot_number
        ].axis("off")


    fig.suptitle(
        f"{name} — Highest-Confidence Validation Errors",
        fontsize=15
    )

    fig.tight_layout()

    fig.savefig(
        output_path,
        dpi=220
    )

    plt.show()
    plt.close(fig)


plot_errors(
    resnet_result,
    "ResNet50",
    EXPLAIN_DIR /
    "ResNet50_misclassified_examples.png"
)

plot_errors(
    effnet_result,
    "EfficientNet-B0",
    EXPLAIN_DIR /
    "EfficientNetB0_misclassified_examples.png"
)


# ================================================================================================
# SECTION 38 — CALIBRATION OF CNNs
# ================================================================================================

fig, ax = plt.subplots(
    figsize=(8, 6)
)

for result in [
    resnet_result,
    effnet_result
]:

    evaluation = result[
        "validation"
    ]

    observed, predicted = (
        calibration_curve(
            evaluation[
                "binary_true"
            ],
            evaluation[
                "fake_probability"
            ],
            n_bins=10,
            strategy="uniform"
        )
    )

    ax.plot(
        predicted,
        observed,
        marker="o",
        label=result[
            "architecture"
        ]
    )


ax.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Perfect Calibration"
)

ax.set_xlabel(
    "Mean Predicted Probability"
)

ax.set_ylabel(
    "Observed Fake Fraction"
)

ax.set_title(
    "CNN Validation Calibration"
)

ax.legend()

fig.tight_layout()

fig.savefig(
    COMPARE_DIR /
    "07_cnn_calibration.png",
    dpi=220
)

plt.show()
plt.close(fig)


# ================================================================================================
# SECTION 39 — DECIDE WHETHER FULL CNN TRAINING IS REQUIRED
# ================================================================================================

PREVIOUS_SWIN_FINAL_TEST = {

    "accuracy":
        0.992800,

    "balanced_accuracy":
        0.992800,

    "precision_fake":
        0.989569,

    "recall_fake":
        0.996100,

    "specificity_real":
        0.989500,

    "f1_fake":
        0.992824,

    "auc_fake":
        0.999774,

    "average_precision_fake":
        0.999773,

    "kappa":
        0.985600,

    "mcc":
        0.985621,

    "brier_fake":
        0.006207
}


PREVIOUS_SWIN_FULL_VALIDATION = {

    "accuracy":
        0.993000,

    "balanced_accuracy":
        0.993000,

    "precision_fake":
        0.989476,

    "recall_fake":
        0.996600,

    "specificity_real":
        0.989400,

    "f1_fake":
        0.993025,

    "auc_fake":
        0.999811,

    "average_precision_fake":
        0.999812,

    "kappa":
        0.986000,

    "mcc":
        0.986026,

    "brier_fake":
        0.006095
}


FINAL_TEST_METRICS = None
FINAL_MODEL_PATH = None
FULL_TRAINING_WAS_REUSED = False


if WINNER == "Swin-Tiny":

    print("\n" + "=" * 110)
    print("SWIN REMAINS THE FOUR-MODEL WINNER")
    print("=" * 110)

    print(
        "✅ No 32-minute Swin retraining will be performed."
    )

    print(
        "The previous completed full Swin training "
        "and untouched-test evaluation are reused."
    )

    FINAL_TEST_METRICS = (
        PREVIOUS_SWIN_FINAL_TEST
    )

    FULL_TRAINING_WAS_REUSED = True


elif WINNER in [
    "ResNet50",
    "EfficientNet-B0"
]:

    print("\n" + "#" * 110)
    print(
        f"{WINNER} DEFEATED SWIN — "
        "FULL CNN TRAINING REQUIRED"
    )
    print("#" * 110)

    if WINNER == "ResNet50":

        cnn_model_name = (
            RESNET_MODEL_NAME
        )

    else:

        cnn_model_name = (
            EFFNET_MODEL_NAME
        )


    full_cnn_result = train_architecture(
        architecture=(
            WINNER +
            "_FULL"
        ),
        model_name=cnn_model_name,
        training_data=train_dataset,
        validation_data=val_dataset,
        epochs=FULL_EPOCHS,
        output_directory=FINAL_DIR,
        keep_model=True
    )


    final_model = full_cnn_result[
        "model"
    ]


    print("\n" + "=" * 110)
    print("FINAL UNTOUCHED TEST EVALUATION")
    print("=" * 110)

    test_loader = make_loader(
        test_dataset,
        EVAL_BATCH_SIZE,
        shuffle=False
    )

    criterion = nn.CrossEntropyLoss()

    test_result = evaluate_model(
        final_model,
        test_loader,
        criterion,
        "FINAL TEST"
    )

    FINAL_TEST_METRICS = (
        test_result[
            "metrics"
        ]
    )


    FINAL_MODEL_PATH = (
        FINAL_DIR /
        "FINAL_CNN_WINNER_STATE_DICT.pt"
    )

    torch.save(
        final_model.state_dict(),
        FINAL_MODEL_PATH
    )


    classification = (
        classification_report(
            test_result["labels"],
            test_result["predictions"],
            labels=[
                FAKE_ID,
                REAL_ID
            ],
            target_names=[
                "Fake",
                "Real"
            ],
            output_dict=True,
            zero_division=0
        )
    )


    pd.DataFrame(
        classification
    ).transpose().to_csv(
        FINAL_DIR /
        "final_test_classification_report.csv"
    )


    plot_cm(
        test_result[
            "confusion_matrix"
        ],
        f"{WINNER} Final Test Confusion Matrix",
        FINAL_DIR /
        "final_test_confusion_matrix.png"
    )


    plot_cm(
        test_result[
            "confusion_matrix"
        ],
        f"{WINNER} Normalized Final Test Confusion Matrix",
        FINAL_DIR /
        "final_test_confusion_matrix_normalized.png",
        normalized=True
    )


else:

    raise RuntimeError(
        "Unexpected winner name."
    )


# ================================================================================================
# SECTION 40 — FINAL RESEARCH TABLE
# ================================================================================================

final_table = comparison_df[
    [
        "model",
        "family",
        "accuracy",
        "balanced_accuracy",
        "precision_fake",
        "recall_fake",
        "specificity_real",
        "f1_fake",
        "auc_fake",
        "average_precision_fake",
        "kappa",
        "mcc",
        "brier_fake",
        "parameters_millions",
        "training_seconds"
    ]
].copy()


final_table.columns = [
    "Model",
    "Family",
    "Accuracy",
    "Balanced Accuracy",
    "Fake Precision",
    "Fake Recall",
    "Real Specificity",
    "Fake F1",
    "ROC-AUC",
    "PR-AUC",
    "Kappa",
    "MCC",
    "Brier Score",
    "Parameters (M)",
    "Training Time (s)"
]


final_table.to_csv(
    COMPARE_DIR /
    "IEEE_READY_FOUR_MODEL_TABLE.csv",
    index=False
)


print("\n" + "=" * 150)
print("IEEE-READY FOUR-MODEL TABLE")
print("=" * 150)

print(
    final_table.to_string(
        index=False
    )
)


# ================================================================================================
# SECTION 41 — FINAL WINNER METRICS
# ================================================================================================

final_metrics_df = pd.DataFrame([
    {
        "winner":
            WINNER,

        "family":
            WINNER_FAMILY,

        "full_training_reused":
            FULL_TRAINING_WAS_REUSED,

        **FINAL_TEST_METRICS
    }
])


final_metrics_df.to_csv(
    FINAL_DIR /
    "FINAL_WINNER_TEST_METRICS.csv",
    index=False
)


print("\n" + "=" * 110)
print("FINAL WINNER TEST PERFORMANCE")
print("=" * 110)

print("Winner:", WINNER)

for metric, value in (
    FINAL_TEST_METRICS.items()
):

    print(
        f"{metric:28s}: "
        f"{value:.6f}"
    )


# ================================================================================================
# SECTION 42 — RESEARCH INSIGHT SUMMARY
# ================================================================================================

best_cnn = (
    comparison_df[
        comparison_df[
            "family"
        ] == "CNN"
    ]
    .sort_values(
        "auc_fake",
        ascending=False
    )
    .iloc[0]
)


best_transformer = (
    comparison_df[
        comparison_df[
            "family"
        ] == "Transformer"
    ]
    .sort_values(
        "auc_fake",
        ascending=False
    )
    .iloc[0]
)


research_insights = f"""
BEYOND ATTENTION — FOUR-MODEL RESEARCH INSIGHTS
================================================================================

1. DATASET AND PREPROCESSING
--------------------------------------------------------------------------------
The study uses the official 140k Real and Fake Faces train/validation/test
structure. Preprocessing is implemented inside this notebook rather than using
precomputed features. All images are converted to RGB, standardized to
224 x 224 input geometry, normalized using ImageNet statistics, and training
images receive controlled stochastic augmentation. Validation and test
transformations remain deterministic.

2. CONTROLLED ARCHITECTURE COMPARISON
--------------------------------------------------------------------------------
The four compared architectures are:
- ViT-Base/16
- Swin-Tiny
- ResNet50
- EfficientNet-B0

All architecture-selection results correspond to the same 10,000-image
training subset and 2,000-image validation subset protocol with seed {SEED}.
ViT and Swin results are reused from the completed earlier experiment rather
than retrained.

3. BEST TRANSFORMER
--------------------------------------------------------------------------------
Model:
{best_transformer["model"]}

Validation ROC-AUC:
{best_transformer["auc_fake"]:.6f}

Validation Accuracy:
{best_transformer["accuracy"]:.6f}

4. BEST CNN
--------------------------------------------------------------------------------
Model:
{best_cnn["model"]}

Validation ROC-AUC:
{best_cnn["auc_fake"]:.6f}

Validation Accuracy:
{best_cnn["accuracy"]:.6f}

5. OVERALL WINNER
--------------------------------------------------------------------------------
Model:
{WINNER}

Family:
{WINNER_FAMILY}

Validation ROC-AUC:
{WINNER_AUC:.6f}

6. FINAL TEST
--------------------------------------------------------------------------------
Accuracy:
{FINAL_TEST_METRICS["accuracy"]:.6f}

Fake F1:
{FINAL_TEST_METRICS["f1_fake"]:.6f}

ROC-AUC:
{FINAL_TEST_METRICS["auc_fake"]:.6f}

MCC:
{FINAL_TEST_METRICS["mcc"]:.6f}

7. SCIENTIFIC INTERPRETATION
--------------------------------------------------------------------------------
The results characterize performance on the supplied real-versus-GAN-generated
face dataset and the defined experimental protocol. They must not be interpreted
as universal performance on all deepfakes, face swaps, diffusion-generated
imagery, manipulated video, or unseen generation pipelines.

The four pretrained architectures also differ in parameter count, inductive
bias, and pretraining history. Therefore, this experiment should be described
as a comparison of practical pretrained architecture pipelines rather than a
perfectly isolated architecture-only ablation.
""".strip()


with open(
    OUTPUT_ROOT /
    "RESEARCH_INSIGHTS.txt",
    "w"
) as file:

    file.write(
        research_insights
    )


print(
    "\n" +
    research_insights
)


# ================================================================================================
# SECTION 43 — SAVE COMPLETE JSON
# ================================================================================================

research_json = {

    "project":
        "Beyond Attention: CNNs vs Vision Transformers for GAN-Generated Face Detection",

    "dataset":
        "140k Real and Fake Faces",

    "dataset_root":
        str(DATASET_ROOT),

    "dataset_sizes": {

        "train":
            len(raw_train),

        "validation":
            len(raw_val),

        "test":
            len(raw_test),

        "total":
            (
                len(raw_train) +
                len(raw_val) +
                len(raw_test)
            )
    },

    "seed":
        SEED,

    "image_size":
        IMAGE_SIZE,

    "bakeoff": {

        "training_images":
            BAKE_TRAIN_SIZE,

        "validation_images":
            BAKE_VAL_SIZE,

        "epochs":
            BAKE_EPOCHS
    },

    "models":
        comparison_df.to_dict(
            orient="records"
        ),

    "winner":
        WINNER,

    "winner_family":
        WINNER_FAMILY,

    "selection_metric":
        "validation fake-class ROC-AUC",

    "full_training_reused":
        FULL_TRAINING_WAS_REUSED,

    "final_test_metrics":
        FINAL_TEST_METRICS
}


with open(
    OUTPUT_ROOT /
    "FOUR_MODEL_RESEARCH_RESULTS.json",
    "w"
) as file:

    json.dump(
        research_json,
        file,
        indent=2
    )


# ================================================================================================
# SECTION 44 — ZIP ALL OUTPUTS
# ================================================================================================

ZIP_PATH = shutil.make_archive(
    "/kaggle/working/"
    "beyond_attention_four_model_research_complete",
    "zip",
    root_dir=str(
        OUTPUT_ROOT
    )
)


# ================================================================================================
# SECTION 45 — COMPLETE
# ================================================================================================

print("\n" + "=" * 120)
print("✅ FOUR-MODEL RESEARCH PIPELINE COMPLETE")
print("=" * 120)

print("\n🏆 FOUR-MODEL WINNER:")
print(WINNER)

print("\nWINNER FAMILY:")
print(WINNER_FAMILY)

print("\nFOUR-MODEL TABLE:")
print(
    COMPARE_DIR /
    "IEEE_READY_FOUR_MODEL_TABLE.csv"
)

print("\nRESEARCH RESULTS JSON:")
print(
    OUTPUT_ROOT /
    "FOUR_MODEL_RESEARCH_RESULTS.json"
)

print("\nRESEARCH INSIGHTS:")
print(
    OUTPUT_ROOT /
    "RESEARCH_INSIGHTS.txt"
)

print("\nALL OUTPUTS:")
print(OUTPUT_ROOT)

print("\nZIP:")
print(ZIP_PATH)

print(
    "\nGenerated research material includes:\n"
    " • dataset class distribution\n"
    " • split leakage check\n"
    " • image format/mode analysis\n"
    " • image resolution analysis\n"
    " • aspect-ratio analysis\n"
    " • brightness/color statistics\n"
    " • sampled corrupt-image audit\n"
    " • sampled exact-duplicate audit\n"
    " • before/after preprocessing visualization\n"
    " • augmentation visualization\n"
    " • ResNet50 training\n"
    " • EfficientNet-B0 training\n"
    " • CNN confusion matrices\n"
    " • CNN probability distributions\n"
    " • CNN ROC curves\n"
    " • CNN precision-recall curves\n"
    " • CNN calibration\n"
    " • four-model performance comparison\n"
    " • model-size comparison\n"
    " • training-time comparison\n"
    " • CNN-vs-Transformer family comparison\n"
    " • Grad-CAM for ResNet50\n"
    " • Grad-CAM for EfficientNet-B0\n"
    " • CNN misclassification analysis\n"
    " • final four-model winner\n"
    " • final test metrics\n"
    " • IEEE-ready results CSV\n"
    " • research JSON\n"
    " • complete ZIP archive"
)

print("\n" + "=" * 120)
print("END")
print("=" * 120)
```
